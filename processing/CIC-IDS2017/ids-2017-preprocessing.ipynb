# Dataset

```python
import numpy as np
import pandas as pd
import os
import glob

for dirname, _, filenames in os.walk('/kaggle/input'):
    for filename in filenames:
        if filename.endswith(".parquet"):
            print(os.path.join(dirname, filename))
```

```python
files = glob.glob("/kaggle/input/**/*.parquet", recursive=True)

# Only files belonging to this CICIDS2017 dataset
expected_names = [
    "Benign-Monday-no-metadata.parquet",
    "Botnet-Friday-no-metadata.parquet",
    "Bruteforce-Tuesday-no-metadata.parquet",
    "DDoS-Friday-no-metadata.parquet",
    "DoS-Wednesday-no-metadata.parquet",
    "Infiltration-Thursday-no-metadata.parquet",
    "Portscan-Friday-no-metadata.parquet",
    "WebAttacks-Thursday-no-metadata.parquet"
]

files = [
    f for f in files
    if os.path.basename(f) in expected_names
]

files = sorted(files)

print("Total files:", len(files))

for f in files:
    print(f)
```

```python
df_temp = pd.read_parquet(files[0])

print("Shape:", df_temp.shape)
df_temp.head()
```

```python
print(df_temp.count())
print("\nTotal columns:", len(df_temp.columns))
```

```python
dfs = []

for file in files:
    print("=" * 80)
    print("Reading:", os.path.basename(file))

    temp = pd.read_parquet(file)

    print("Shape:", temp.shape)

    dfs.append(temp)
```

```python
df = pd.concat(dfs, ignore_index=True)

print("Combined dataset shape:", df.shape)
```

```python
df.info()
```

```python
df.describe()
```

# Preprocessing

## 1. Missing Value Check

```python
missing = df.isnull().sum()

print(missing[missing > 0])
print("\nTotal missing values:", missing.sum())
```

## 2. Infinite Value Check

```python
numeric_cols = df.select_dtypes(include=np.number).columns

inf_count = np.isinf(df[numeric_cols]).sum()

print(inf_count[inf_count > 0])
print("\nTotal infinity values:", inf_count.sum())
```

## 3. Data Cleaning

```python
df.replace([np.inf, -np.inf], np.nan, inplace=True)
```

```python
print("Rows before cleaning:", len(df))

df = df.dropna().reset_index(drop=True)

print("Rows after cleaning:", len(df))
```

## 4. Duplicate Removal

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates().reset_index(drop=True)

print("Shape after removing duplicates:", df.shape)
```

```python
df.describe()
```

## 5. Flow Duration Validation

```python
if "Flow Duration" in df.columns:

    print("Negative Flow Duration:",
          (df["Flow Duration"] < 0).sum())

    df = df[df["Flow Duration"] >= 0].reset_index(drop=True)
```

## 6. Number of Target Classes

```python
print("Number of classes:", df["Label"].nunique())

print("\nClasses:")

for label in sorted(df["Label"].astype(str).unique()):
    print(label)
```

## 7. Class Distribution Analysis

```python
class_counts = df["Label"].value_counts()

print(class_counts)
```

```python
df_before = df.copy()
```

## 8. Label Encoding

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df["Label"] = le.fit_transform(
    df["Label"].astype(str)
)

print("Class Mapping:")

for i, class_name in enumerate(le.classes_):
    print(i, "->", class_name)
```

## 9. Encoded Class Distribution

```python
print(df["Label"].value_counts().sort_index())
```

## 10. Feature-Target Separation

```python
X = df.drop("Label", axis=1)
y = df["Label"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

## 11. Train-Test Split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=42
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

## 12. Min-Max Scaling

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("X_train scaled shape:", X_train_scaled.shape)
print("X_test scaled shape:", X_test_scaled.shape)
```

```python
# Save the scaled train/test data

pd.DataFrame(
    X_train_scaled,
    columns=X_train.columns
).to_parquet("X_train_scaled.parquet", index=False)

pd.DataFrame(
    X_test_scaled,
    columns=X_test.columns
).to_parquet("X_test_scaled.parquet", index=False)

# Save labels
y_train.to_frame("Label").to_parquet("y_train.parquet", index=False)
y_test.to_frame("Label").to_parquet("y_test.parquet", index=False)
```

## 13. Save the train/test data

```python
X_train.to_parquet("X_train.parquet", index=False)
X_test.to_parquet("X_test.parquet", index=False)

y_train.to_frame("Label").to_parquet("y_train.parquet", index=False)
y_test.to_frame("Label").to_parquet("y_test.parquet", index=False)
```
