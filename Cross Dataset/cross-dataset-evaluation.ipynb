```python
def map_2017_label(label):
    label = str(label).strip()

    mapping = {
        "Benign": "Benign",
        "Bot": "Bot",

        "FTP-Patator": "FTP",
        "SSH-Patator": "SSH",

        "Web Attack � Brute Force": "Brute Force",
        "Web Attack � XSS": "XSS",
        "Web Attack � Sql Injection": "SQL Injection",

        "DoS slowloris": "DoS",
        "DoS Slowhttptest": "DoS",
        "DoS Hulk": "DoS",
        "DoS GoldenEye": "DoS",

        "DDoS": "DDoS",

        "Infiltration": "Infiltration",

        # No corresponding 2018 model class
        "PortScan": "Other",
        "Heartbleed": "Other"
    }

    return mapping.get(label, "Other")


y_true_harmonized = y_cross_original.apply(map_2017_label)

print("2017 harmonized labels:")
print(y_true_harmonized.value_counts())
```

```python
import joblib

xgb = joblib.load(
    "/kaggle/input/notebooks/hafsaferdousi5/2-2xgboost/outputs/models/XGboost.pkl"
)

le = joblib.load(
    "/kaggle/input/notebooks/hafsaferdousi5/01-data-preprocessing/outputs/models/label_encoder.pkl"
)

scaler = joblib.load(
    "/kaggle/input/notebooks/hafsaferdousi5/01-data-preprocessing/outputs/models/minmax_scaler.pkl"
)

feature_columns = joblib.load(
    "/kaggle/input/notebooks/hafsaferdousi5/3-feature-selection/outputs/features/feature_names.pkl"
)

print("Label encoder:", type(le))
print("Scaler:", type(scaler))
print("Features:", len(feature_columns))
```

```python
X_cross = df[feature_columns].copy()
y_cross_original = df["Label"].copy()

X_cross = X_cross.replace([np.inf, -np.inf], np.nan)
X_cross = X_cross.fillna(0)

X_cross_scaled = scaler.transform(X_cross)

print("Cross-dataset test shape:", X_cross_scaled.shape)
```

```python
models = {
    "XGBoost": xgb,
    "Random Forest": rf,
    "LightGBM": lgbm,
    "CatBoost": catb
}

predictions = {}

for name, model in models.items():

    print(f"\nRunning {name}...")

    pred_encoded = model.predict(X_cross_scaled)

    # Handle models that return shape (n, 1)
    pred_encoded = np.asarray(pred_encoded).reshape(-1)

    predictions[name] = le.inverse_transform(
        pred_encoded.astype(int)
    )

    print(f"{name} completed.")
    print("Predicted classes:", np.unique(predictions[name]))
```

```python
def map_2018_label(label):
    label = str(label).strip()

    mapping = {
        "Benign": "Benign",
        "Bot": "Bot",

        "FTP-BruteForce": "FTP",
        "SSH-Bruteforce": "SSH",

        "Brute Force -Web": "Brute Force",
        "Brute Force -XSS": "XSS",

        "SQL Injection": "SQL Injection",

        "DoS attacks-GoldenEye": "DoS",
        "DoS attacks-Hulk": "DoS",
        "DoS attacks-SlowHTTPTest": "DoS",
        "DoS attacks-Slowloris": "DoS",

        "DDOS attack-HOIC": "DDoS",
        "DDOS attack-LOIC-UDP": "DDoS",
        "DDoS attacks-LOIC-HTTP": "DDoS",

        "Infilteration": "Infiltration"
    }

    return mapping.get(label, "Other")
```

```python
predictions_harmonized = {}

for name, preds in predictions.items():

    predictions_harmonized[name] = pd.Series(preds).apply(
        map_2018_label
    )

    print(f"\n{name} prediction labels:")
    print(predictions_harmonized[name].value_counts())
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef
)

# Actual labels
y_true_harmonized = y_true_harmonized.reset_index(drop=True)

# Keep only samples with labels represented in the 2018 model
common_mask = y_true_harmonized != "Other"

y_true_common = y_true_harmonized[
    common_mask
].reset_index(drop=True)

results = []

for name, preds in predictions_harmonized.items():

    preds = preds.reset_index(drop=True)

    y_pred_common = preds[
        common_mask
    ].reset_index(drop=True)

    accuracy = accuracy_score(
        y_true_common,
        y_pred_common
    )

    precision = precision_score(
        y_true_common,
        y_pred_common,
        average="weighted",
        zero_division=0
    )

    recall = recall_score(
        y_true_common,
        y_pred_common,
        average="weighted",
        zero_division=0
    )

    f1 = f1_score(
        y_true_common,
        y_pred_common,
        average="weighted",
        zero_division=0
    )

    # Matthews Correlation Coefficient
    mcc = matthews_corrcoef(
        y_true_common,
        y_pred_common
    )

    results.append({
        "Model": name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-score": f1,
        "MCC": mcc
    })

results_df = pd.DataFrame(results)

print("\n==============================================")
print(" CROSS-DATASET MULTICLASS RESULTS")
print(" Train: CSE-CIC-IDS2018")
print(" Test : CIC-IDS2017")
print("==============================================")

display(
    results_df.style.format({
        "Accuracy": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1-score": "{:.4f}",
        "MCC": "{:.4f}"
    })
)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef,
    roc_auc_score
)

# Actual labels
y_true_harmonized = y_true_harmonized.reset_index(drop=True)

# Keep only samples with labels represented in the 2018 model
common_mask = y_true_harmonized != "Other"

y_true_common = y_true_harmonized[
    common_mask
].reset_index(drop=True)

results = []

for name, preds in predictions_harmonized.items():

    # Predicted labels
    preds = preds.reset_index(drop=True)

    y_pred_common = preds[
        common_mask
    ].reset_index(drop=True)

    # -----------------------------
    # Classification metrics
    # -----------------------------
    accuracy = accuracy_score(
        y_true_common,
        y_pred_common
    )

    precision = precision_score(
        y_true_common,
        y_pred_common,
        average="weighted",
        zero_division=0
    )

    recall = recall_score(
        y_true_common,
        y_pred_common,
        average="weighted",
        zero_division=0
    )

    f1 = f1_score(
        y_true_common,
        y_pred_common,
        average="weighted",
        zero_division=0
    )

    # -----------------------------
    # Matthews Correlation Coefficient
    # -----------------------------
    mcc = matthews_corrcoef(
        y_true_common,
        y_pred_common
    )

    # -----------------------------
    # Multiclass ROC-AUC
    # -----------------------------
    model = models[name]

    y_proba = model.predict_proba(X_cross_scaled)

    # Apply the same mask used for evaluation
    y_proba_common = y_proba[
        common_mask.values
    ]

    # Model classes -> original labels
    model_classes = le.inverse_transform(
        np.arange(len(model.classes_))
    )

    # Harmonize model class names
    model_classes_harmonized = np.array([
        map_2018_label(label)
        for label in model_classes
    ])

    # Get the classes actually present after harmonization
    auc_classes = np.unique(
        y_true_common
    )

    # Create probability matrix for the common classes
    auc_proba = np.zeros(
        (len(y_true_common), len(auc_classes))
    )

    for i, cls in enumerate(auc_classes):

        matching = np.where(
            model_classes_harmonized == cls
        )[0]

        if len(matching) > 0:
            auc_proba[:, i] = y_proba_common[:, matching[0]]

    # Normalize probabilities
    row_sums = auc_proba.sum(axis=1)
    valid_rows = row_sums > 0

    auc_proba[valid_rows] = (
        auc_proba[valid_rows]
        / row_sums[valid_rows, None]
    )

    # Encode true labels
    y_true_auc = pd.Categorical(
        y_true_common,
        categories=auc_classes
    ).codes

    # Calculate weighted multiclass ROC-AUC
    try:
        auc = roc_auc_score(
            y_true_auc,
            auc_proba,
            multi_class="ovr",
            average="weighted",
            labels=np.arange(len(auc_classes))
        )
    except ValueError:
        auc = np.nan

    # -----------------------------
    # Store results
    # -----------------------------
    results.append({
        "Model": name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-score": f1,
        "MCC": mcc,
        "AUC": auc
    })


# Final results table
results_df = pd.DataFrame(results)

print("\n==============================================")
print(" CROSS-DATASET MULTICLASS RESULTS")
print(" Train: CSE-CIC-IDS2018")
print(" Test : CIC-IDS2017")
print("==============================================")

display(
    results_df.style.format({
        "Accuracy": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1-score": "{:.4f}",
        "MCC": "{:.4f}",
        "AUC": "{:.4f}"
    })
)
```
